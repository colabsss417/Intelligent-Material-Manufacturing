# PREPROCESSING

```python
# ============================================================
# STEP 2 + STEP 3 + STEP 4
# DATA CLEANING + KNN IMPUTATION + Z-SCORE STANDARDIZATION
# ============================================================

import os
import numpy as np
import pandas as pd

from sklearn.impute import KNNImputer
from sklearn.preprocessing import StandardScaler


# ============================================================
# PATHS
# ============================================================

input_path = r"E:\Projects\1_Material Manufacturing\Dataset\C_Dataset.csv"

output_path = (
    r"E:\Projects\1_Material Manufacturing\Dataset\C_Dataset_preprocessed.csv"
)


# ============================================================
# STEP 2: DATA CLEANING AND QUALITY CHECKING
# ============================================================

print("=" * 90)
print(" DATA CLEANING")
print("=" * 90)

# Check input file
if not os.path.exists(input_path):
    raise FileNotFoundError(
        f"\nInput dataset was not found:\n{input_path}"
    )

# Load dataset
df = pd.read_csv(input_path)


# ------------------------------------------------------------
# 2.1 Remove completely empty rows
# ------------------------------------------------------------

empty_rows = df.isna().all(axis=1).sum()

if empty_rows > 0:
    df = df.dropna(how="all").reset_index(drop=True)

print(f"\nCompletely empty rows removed: {empty_rows}")


# ------------------------------------------------------------
# 2.2 Remove completely empty columns
# ------------------------------------------------------------

empty_columns = df.columns[df.isna().all()].tolist()

if len(empty_columns) > 0:
    df = df.drop(columns=empty_columns)

print(f"Completely empty columns removed: {len(empty_columns)}")

if empty_columns:
    print("Removed columns:")
    for col in empty_columns:
        print("  -", col)


# ------------------------------------------------------------
# 2.3 Remove duplicate records
# ------------------------------------------------------------

duplicate_count = df.duplicated().sum()

if duplicate_count > 0:
    df = df.drop_duplicates().reset_index(drop=True)


# ------------------------------------------------------------
# 2.4 Clean column names
# ------------------------------------------------------------

df.columns = (
    df.columns
    .astype(str)
    .str.strip()
    .str.replace(r"\s+", "_", regex=True)
)


# ------------------------------------------------------------
# 2.5 Convert common missing-value representations to NaN
# ------------------------------------------------------------

missing_tokens = [
    "",
    " ",
    "NA",
    "N/A",
    "na",
    "n/a",
    "NULL",
    "null",
    "None",
    "none",
    "?",
    "-",
    "--"
]

df = df.replace(missing_tokens, np.nan)


# ------------------------------------------------------------
# 2.6 Convert numeric-looking object columns to numeric
# ------------------------------------------------------------

for col in df.columns:

    if df[col].dtype == "object":

        converted = pd.to_numeric(
            df[col],
            errors="coerce"
        )

        # Convert only when most non-null values are numeric
        non_null_count = df[col].notna().sum()

        if non_null_count > 0:

            numeric_count = converted.notna().sum()

            if numeric_count / non_null_count >= 0.80:
                df[col] = converted


# ------------------------------------------------------------
# 2.7 Identify target column
# ------------------------------------------------------------

target_candidates = [
    "IEC_Alignment_Class",
    "Professional_Competency_Index"
]

target_column = None

for col in target_candidates:
    if col in df.columns:
        target_column = col
        break


if target_column is None:

    print("\nWARNING:")
    print("No predefined target column was detected.")
    print("The preprocessing will exclude obvious ID columns")
    print("but will not automatically select a target.")

else:

    print(f"\nTarget column detected: {target_column}")


# ------------------------------------------------------------
# 2.8 Identify ID columns
# ------------------------------------------------------------

id_columns = []

for col in df.columns:

    col_lower = col.lower()

    if (
        col_lower == "id"
        or col_lower.endswith("_id")
        or col_lower.startswith("id_")
        or "student_id" in col_lower
        or "record_id" in col_lower
    ):
        id_columns.append(col)



# ------------------------------------------------------------
# 2.9 Missing-value report before imputation
# ------------------------------------------------------------

missing_before = df.isna().sum()

print("\nMissing values before KNN imputation:")

missing_before_nonzero = missing_before[
    missing_before > 0
]

if len(missing_before_nonzero) == 0:

    print("  No missing values found.")

else:

    print(
        missing_before_nonzero
        .sort_values(ascending=False)
        .to_string()
    )


# ============================================================
# PREPARE FEATURES FOR KNN IMPUTATION
# ============================================================

# Preserve target separately
if target_column is not None:

    target = df[target_column].copy()

    feature_df = df.drop(
        columns=[target_column]
    ).copy()

else:

    target = None
    feature_df = df.copy()


# ------------------------------------------------------------
# Separate ID columns
# ------------------------------------------------------------

existing_id_columns = [
    col for col in id_columns
    if col in feature_df.columns
]

id_data = feature_df[existing_id_columns].copy()

feature_df = feature_df.drop(
    columns=existing_id_columns,
    errors="ignore"
)


# ------------------------------------------------------------
# Identify numeric and categorical predictors
# ------------------------------------------------------------

numeric_columns = feature_df.select_dtypes(
    include=np.number
).columns.tolist()

categorical_columns = feature_df.select_dtypes(
    exclude=np.number
).columns.tolist()


# ============================================================
# STEP 3: KNN IMPUTATION
# ============================================================

print("\n" + "=" * 90)
print(" KNN IMPUTATION")
print("=" * 90)


# ------------------------------------------------------------
# KNN works directly on numerical predictors.
#
# For categorical columns:
# - preserve them unchanged
# - fill missing categorical values using mode
#
# Numerical columns:
# - KNN imputation
# ------------------------------------------------------------

numeric_data = feature_df[numeric_columns].copy()

categorical_data = feature_df[categorical_columns].copy()


# ------------------------------------------------------------
# KNN imputation
# ------------------------------------------------------------

if len(numeric_columns) > 0:

    # Number of neighbors
    K_NEIGHBORS = 5

    print(
        f"\nKNN Imputation parameters:"
        f"\n  Number of neighbors (k): {K_NEIGHBORS}"
    )

    imputer = KNNImputer(
        n_neighbors=K_NEIGHBORS,
        weights="distance"
    )

    numeric_imputed_array = imputer.fit_transform(
        numeric_data
    )

    numeric_imputed = pd.DataFrame(
        numeric_imputed_array,
        columns=numeric_columns,
        index=feature_df.index
    )

else:

    numeric_imputed = pd.DataFrame(
        index=feature_df.index
    )


# ------------------------------------------------------------
# Categorical missing-value handling
# ------------------------------------------------------------

for col in categorical_columns:

    if categorical_data[col].isna().any():

        mode_values = categorical_data[col].mode(
            dropna=True
        )

        if len(mode_values) > 0:

            categorical_data[col] = (
                categorical_data[col]
                .fillna(mode_values.iloc[0])
            )

        else:

            categorical_data[col] = (
                categorical_data[col]
                .fillna("Unknown")
            )


# ------------------------------------------------------------
# Recombine numerical and categorical features
# ------------------------------------------------------------

processed_features = pd.concat(
    [
        numeric_imputed,
        categorical_data
    ],
    axis=1
)

# Restore original feature order
processed_features = processed_features[
    feature_df.columns
]


# ------------------------------------------------------------
# Missing-value verification
# ------------------------------------------------------------

remaining_missing = processed_features.isna().sum().sum()

print(
    f"\nRemaining missing feature values after KNN: "
    f"{remaining_missing}"
)


# ============================================================
# STEP 4: Z-SCORE STANDARDIZATION
# ============================================================

print("\n" + "=" * 90)
print("Z-SCORE STANDARDIZATION")
print("=" * 90)


# ------------------------------------------------------------
# Standardize numerical features
#
# Formula:
#
#       Z = (X - mean) / standard deviation
#
# Categorical features are not standardized.
# ------------------------------------------------------------

if len(numeric_columns) > 0:

    scaler = StandardScaler()

    scaled_array = scaler.fit_transform(
        processed_features[numeric_columns]
    )

    scaled_numeric = pd.DataFrame(
        scaled_array,
        columns=numeric_columns,
        index=processed_features.index
    )

else:

    scaled_numeric = pd.DataFrame(
        index=processed_features.index
    )


# ------------------------------------------------------------
# Combine standardized numerical and categorical features
# ------------------------------------------------------------

final_features = pd.concat(
    [
        scaled_numeric,
        processed_features[categorical_columns]
    ],
    axis=1
)


# Restore original predictor order
final_features = final_features[
    processed_features.columns
]


# ============================================================
# RECONSTRUCT FINAL DATASET
# ============================================================

final_df_parts = []


# ID columns first
if len(existing_id_columns) > 0:

    final_df_parts.append(
        id_data.reset_index(drop=True)
    )


# Processed features
final_df_parts.append(
    final_features.reset_index(drop=True)
)


# Target column last
if target is not None:

    final_df_parts.append(
        target.reset_index(drop=True)
    )


# Combine everything
final_df = pd.concat(
    final_df_parts,
    axis=1
)

# ------------------------------------------------------------
# Verify standardized numerical features
# ------------------------------------------------------------

if len(numeric_columns) > 0:

    verification = pd.DataFrame({
        "Mean_After_Scaling":
            final_df[numeric_columns].mean(),

        "Std_After_Scaling":
            final_df[numeric_columns].std()
    })

    print("\nZ-score verification:")
    print(verification.round(4).to_string())


# ============================================================
# SAVE PREPROCESSED DATASET
# ============================================================

output_directory = os.path.dirname(output_path)

os.makedirs(
    output_directory,
    exist_ok=True
)

final_df.to_csv(
    output_path,
    index=False
)


# ============================================================
# VERIFY SAVED FILE
# ============================================================

if os.path.exists(output_path):

    file_size_mb = (
        os.path.getsize(output_path)
        / (1024 ** 2)
    )

    print("\n" + "=" * 90)
    print("PREPROCESSING COMPLETED SUCCESSFULLY")
    print("=" * 90)

    print(
        f"\nSaved file:\n{output_path}"
    )

    print(
        f"\nFile size: {file_size_mb:.2f} MB"
    )

    print(
        f"Final dataset shape: {final_df.shape}"
    )

else:

    print(
        "\nERROR: Preprocessed file was not created."
    )


# ============================================================
# DISPLAY SAMPLE
# ============================================================

print("\nFirst 5 preprocessed records:")
print(final_df.shape)
print(
    final_df.head()
)
```

# Feature extraction

```python
import os
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA

# ============================================================
# STEP 5: PCA FEATURE EXTRACTION
# ============================================================

# -----------------------------
# File paths
# -----------------------------
input_file = r"E:\Projects\1_Material Manufacturing\Dataset\C_Dataset_preprocessed.csv"

output_file = r"E:\Projects\1_Material Manufacturing\Dataset\C_Dataset_feature_extracted.csv"

# -----------------------------
# Check input file
# -----------------------------
if not os.path.exists(input_file):
    raise FileNotFoundError(
        f"Preprocessed file not found:\n{input_file}\n\n"
        "Please make sure Step 2-4 was completed successfully."
    )

# -----------------------------
# Load dataset
# -----------------------------
df = pd.read_csv(input_file)

print("=" * 80)
print(" PCA FEATURE EXTRACTION")
print("=" * 80)

# ============================================================
# 1. IDENTIFY TARGET COLUMN
# ============================================================

target_candidates = [
    "IEC_Alignment_Class",
    "Professional_Competency_Index"
]

target_col = None

for col in target_candidates:
    if col in df.columns:
        target_col = col
        break

if target_col is None:
    raise ValueError(
        "Target column not found.\n"
        "Expected one of:\n"
        f"{target_candidates}"
    )


# ============================================================
# 2. IDENTIFY ID COLUMNS
# ============================================================

id_cols = []

for col in df.columns:
    col_lower = col.lower()

    if (
        col_lower == "id"
        or col_lower.endswith("_id")
        or col_lower.startswith("id_")
        or "student_id" in col_lower
        or "record_id" in col_lower
    ):
        if col != target_col:
            id_cols.append(col)

# ============================================================
# 3. SEPARATE TARGET
# ============================================================

target = df[target_col].copy()

# ============================================================
# 4. SELECT NUMERICAL PREDICTOR FEATURES
# ============================================================

exclude_cols = id_cols + [target_col]

feature_df = df.drop(columns=exclude_cols, errors="ignore")

# Select only numerical columns
numeric_features = feature_df.select_dtypes(
    include=[np.number]
).columns.tolist()

if len(numeric_features) < 15:
    raise ValueError(
        f"\nOnly {len(numeric_features)} numerical predictor features "
        "are available.\n"
        "PCA cannot produce 20 components. "
        "Please check the preprocessed dataset."
    )

X = feature_df[numeric_features].copy()

# ============================================================
# 5. CHECK MISSING VALUES
# ============================================================

missing_total = X.isna().sum().sum()

print("\nTotal missing values in PCA input:")
print(missing_total)

if missing_total > 0:
    raise ValueError(
        "Missing values are still present in the PCA input. "
        "Please complete Step 3 KNN imputation before Step 5."
    )

# ============================================================
# 6. CHECK INFINITE VALUES
# ============================================================

if np.isinf(X.to_numpy()).any():
    raise ValueError(
        "Infinite values detected in PCA input. "
        "Please clean the preprocessed dataset first."
    )

# ============================================================
# 7. PCA FEATURE EXTRACTION
# ============================================================

n_components = 15

print("\nApplying PCA...")
print(f"Requested PCA components: {n_components}")

pca = PCA(
    n_components=n_components,
    random_state=42
)

X_pca = pca.fit_transform(X)

# ============================================================
# 8. CREATE PCA FEATURE DATAFRAME
# ============================================================

pca_columns = [
    f"PCA_{i}"
    for i in range(1, n_components + 1)
]

pca_df = pd.DataFrame(
    X_pca,
    columns=pca_columns,
    index=df.index
)

# ============================================================
# 9. RECONSTRUCT FINAL DATASET
# ============================================================

final_parts = []

# Keep ID columns
if id_cols:
    final_parts.append(df[id_cols].reset_index(drop=True))

# Add PCA features
final_parts.append(
    pca_df.reset_index(drop=True)
)

# Add target
final_parts.append(
    target.reset_index(drop=True).to_frame()
)

feature_extracted_df = pd.concat(
    final_parts,
    axis=1
)

# ============================================================
# 10. PCA EXPLAINED VARIANCE
# ============================================================

explained_variance = pca.explained_variance_ratio_

cumulative_variance = np.cumsum(
    explained_variance
)

variance_df = pd.DataFrame({
    "Component": pca_columns,
    "Explained_Variance_Ratio": explained_variance,
    "Explained_Variance_Percentage": explained_variance * 100,
    "Cumulative_Variance_Percentage": cumulative_variance * 100
})

# ============================================================
# 11. DISPLAY PCA RESULTS
# ============================================================

print("\n" + "=" * 80)
print("PCA RESULTS")
print("=" * 80)

print("\nPCA feature shape:")
print(X_pca.shape)

print("\nExplained variance by component:")
print(variance_df.to_string(index=False))

print("\nTotal variance explained by 20 PCA components:")
print(f"{cumulative_variance[-1] * 100:.2f}%")

# ============================================================
# 12. CHECK FINAL DATASET
# ============================================================

print("\n" + "=" * 80)
print("FINAL FEATURE-EXTRACTED DATASET")
print("=" * 80)

print("\nShape:")
print(feature_extracted_df.shape)

print("\nFirst 5 rows:")
print(feature_extracted_df.head())

# ============================================================
# 14. SAVE FEATURE-EXTRACTED CSV
# ============================================================

feature_extracted_df.to_csv(
    output_file,
    index=False
)

print("\n" + "=" * 80)
print("FEATURE-EXTRACTION COMPLETED")
print("=" * 80)
```

# SEQUENCE CONSTRUCTION

```python
import os
import numpy as np
import pandas as pd

# ============================================================
# STEP 6: FIXED-LENGTH SEQUENCE CONSTRUCTION
# OUTPUT: ONE CSV FILE ONLY
# ============================================================

INPUT_FILE = (
    r"E:\Projects\1_Material Manufacturing\Dataset"
    r"\C_Dataset_feature_extracted.csv"
)

OUTPUT_FILE = (
    r"E:\Projects\1_Material Manufacturing\Dataset"
    r"\C_Dataset_sequence_feature_extracted.csv"
)

SEQUENCE_LENGTH = 50
TARGET_COL = "IEC_Alignment_Class"

# ============================================================
# 1. LOAD DATASET
# ============================================================

if not os.path.exists(INPUT_FILE):
    raise FileNotFoundError(
        f"Input file not found:\n{INPUT_FILE}"
    )

df = pd.read_csv(INPUT_FILE)

print("=" * 80)
print("STEP 6: FIXED-LENGTH SEQUENCE CONSTRUCTION")
print("=" * 80)

print("\nInput file:")
print(INPUT_FILE)

print("\nInput shape:")
print(df.shape)

print("\nColumns:")
print(df.columns.tolist())

# ============================================================
# 2. IDENTIFY STUDENT ID
# ============================================================

student_candidates = [
    "Student_ID",
    "student_id",
    "StudentID",
    "studentid"
]

student_col = None

for col in student_candidates:
    if col in df.columns:
        student_col = col
        break

if student_col is None:
    for col in df.columns:
        if "student" in col.lower() and "id" in col.lower():
            student_col = col
            break

if student_col is None:
    raise ValueError(
        "Student ID column was not found."
    )

print("\nStudent ID column:")
print(student_col)

# ============================================================
# 3. AUTOMATICALLY DETECT PCA FEATURES
# ============================================================

pca_features = []

for col in df.columns:

    if col.startswith("PCA_"):

        try:
            number = int(col.replace("PCA_", ""))
            pca_features.append((number, col))

        except ValueError:
            pass

pca_features = [
    col
    for number, col in sorted(
        pca_features,
        key=lambda x: x[0]
    )
]

if len(pca_features) == 0:
    raise ValueError(
        "No PCA features were found."
    )

print("\nPCA features detected:")
print(pca_features)

print("\nNumber of PCA features:")
print(len(pca_features))

# ============================================================
# 4. TARGET CHECK
# ============================================================

if TARGET_COL not in df.columns:
    raise ValueError(
        f"Target column '{TARGET_COL}' not found."
    )

# ============================================================
# 5. CHECK MISSING VALUES
# ============================================================

required_columns = (
    [student_col]
    + pca_features
    + [TARGET_COL]
)

missing_values = (
    df[required_columns]
    .isna()
    .sum()
    .sum()
)

print("\nMissing values:")
print(missing_values)

if missing_values > 0:
    raise ValueError(
        "Missing values detected. "
        "Please complete preprocessing before Step 6."
    )

# ============================================================
# 6. OBSERVATIONS PER STUDENT
# ============================================================

student_counts = (
    df.groupby(student_col)
      .size()
      .sort_values()
)

print("\n" + "=" * 80)
print("STUDENT OBSERVATION DISTRIBUTION")
print("=" * 80)

print("\nNumber of students:")
print(len(student_counts))

print("\nObservation statistics:")
print(student_counts.describe())

print("\nStudents with exactly 50 observations:")
print(
    (student_counts == SEQUENCE_LENGTH).sum()
)

print("\nStudents with fewer than 50 observations:")
print(
    (student_counts < SEQUENCE_LENGTH).sum()
)

print("\nStudents with more than 50 observations:")
print(
    (student_counts > SEQUENCE_LENGTH).sum()
)

# ============================================================
# 7. CREATE SEQUENCES
# ============================================================

sequence_rows = []

sequence_id = 1

valid_students = student_counts[
    student_counts >= SEQUENCE_LENGTH
].index

for student_id in valid_students:

    student_data = df[
        df[student_col] == student_id
    ].copy()

    # Preserve original row order
    student_data = student_data.reset_index(drop=True)

    n_observations = len(student_data)

    # Number of complete 50-observation sequences
    n_sequences = (
        n_observations // SEQUENCE_LENGTH
    )

    for seq_idx in range(n_sequences):

        start = (
            seq_idx
            * SEQUENCE_LENGTH
        )

        end = (
            start
            + SEQUENCE_LENGTH
        )

        sequence = student_data.iloc[
            start:end
        ]

        if len(sequence) != SEQUENCE_LENGTH:
            continue

        # ----------------------------------------------------
        # Sequence-level target
        # ----------------------------------------------------
        #
        # Uses the target of the final observation.
        # ----------------------------------------------------

        target = sequence[
            TARGET_COL
        ].iloc[-1]

        # ----------------------------------------------------
        # Create ONE ROW for each sequence
        # ----------------------------------------------------
        #
        # PCA features are flattened:
        #
        # PCA_1_Obs_1 ... PCA_15_Obs_1
        # PCA_1_Obs_2 ... PCA_15_Obs_2
        # ...
        # PCA_1_Obs_50 ... PCA_15_Obs_50
        #
        # ----------------------------------------------------

        row = {
            "Sequence_ID": sequence_id,
            "Student_ID": student_id,
            "Sequence_Index": seq_idx + 1,
            "Sequence_Length": SEQUENCE_LENGTH
        }

        # Add every PCA feature for every observation
        for obs_num in range(SEQUENCE_LENGTH):

            observation = sequence.iloc[obs_num]

            for feature in pca_features:

                column_name = (
                    f"{feature}_Obs_{obs_num + 1}"
                )

                row[column_name] = (
                    observation[feature]
                )

        # Add target at the end
        row[TARGET_COL] = target

        sequence_rows.append(row)

        sequence_id += 1

# ============================================================
# 8. CREATE FINAL SEQUENCE DATAFRAME
# ============================================================

sequence_df = pd.DataFrame(
    sequence_rows
)

# ============================================================
# 9. CHECK RESULTS
# ============================================================

print("\n" + "=" * 80)
print("SEQUENCE CONSTRUCTION RESULTS")
print("=" * 80)

print("\nNumber of sequences:")
print(len(sequence_df))

print("\nFinal CSV shape:")
print(sequence_df.shape)

print("\nNumber of PCA features per observation:")
print(len(pca_features))

print("\nSequence length:")
print(SEQUENCE_LENGTH)

print("\nTotal sequence feature values:")
print(
    len(pca_features)
    * SEQUENCE_LENGTH
)

# ============================================================
# 10. VERIFY EXPECTED NUMBER OF COLUMNS
# ============================================================

expected_columns = (
    4
    + (
        len(pca_features)
        * SEQUENCE_LENGTH
    )
    + 1
)

print("\nExpected number of columns:")
print(expected_columns)

print("Actual number of columns:")
print(sequence_df.shape[1])

# ============================================================
# 11. CHECK MISSING VALUES
# ============================================================

final_missing = (
    sequence_df
    .isna()
    .sum()
    .sum()
)

print("\nMissing values in final dataset:")
print(final_missing)

# ============================================================
# 12. CHECK DUPLICATES
# ============================================================

duplicates = sequence_df.duplicated().sum()

print("\nDuplicate sequences:")
print(duplicates)

# ============================================================
# 13. TARGET DISTRIBUTION
# ============================================================

print("\n" + "=" * 80)
print("TARGET DISTRIBUTION")
print("=" * 80)

print(
    sequence_df[TARGET_COL]
    .value_counts()
    .sort_index()
)

# ============================================================
# 14. SAVE ONE CSV ONLY
# ============================================================

sequence_df.to_csv(
    OUTPUT_FILE,
    index=False
)

# ============================================================
# 15. VERIFY SAVED FILE
# ============================================================

if os.path.exists(OUTPUT_FILE):

    saved_df = pd.read_csv(
        OUTPUT_FILE,
        nrows=5
    )

    print("\n" + "=" * 80)
    print("STEP 6 COMPLETED SUCCESSFULLY")
    print("=" * 80)

    print("\nSaved file:")
    print(OUTPUT_FILE)

    print("\nFinal dataset shape:")
    print(sequence_df.shape)

    print("\nSaved file verified:")
    print("YES")

    print("\nFirst 5 rows:")
    print(saved_df.head())

else:

    raise IOError(
        "The output CSV could not be created."
    )

print("\n" + "=" * 80)
print("ONLY ONE OUTPUT FILE WAS CREATED")
print("=" * 80)

print(OUTPUT_FILE)
```

```python
# ============================================================
# STEP 7
# PROPOSED RES-ATTRRNN MODEL
# COMPLETE RESULT PRINTING
# ============================================================

import os
import time
import random
import warnings

import numpy as np
import pandas as pd
import tensorflow as tf

from sklearn.model_selection import train_test_split

from tensorflow.keras import Model, Input
from tensorflow.keras.layers import (
    Dense,
    Dropout,
    Add,
    SimpleRNN,
    Layer,
    TimeDistributed,
    Concatenate
)
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau
)
import warnings
warnings.filterwarnings("ignore")


# ============================================================
# 1. PATH
# ============================================================

CSV_PATH = (
    r"E:\Projects\1_Material Manufacturing\Dataset\C_Dataset_sequence_feature_extracted.csv"
)


# ============================================================
# 2. REPRODUCIBILITY
# ============================================================

GLOBAL_SEED = 42

os.environ["PYTHONHASHSEED"] = str(GLOBAL_SEED)

random.seed(GLOBAL_SEED)
np.random.seed(GLOBAL_SEED)
tf.random.set_seed(GLOBAL_SEED)


# ============================================================
# 3. LOAD DATA
# ============================================================

df = pd.read_csv(CSV_PATH)

print("\n" + "=" * 80)
print("PROPOSED RES-ATTRRNN MODEL")
print("=" * 80)


# ============================================================
# 4. IDENTIFY PCA FEATURES
# ============================================================

pca_cols = [
    c for c in df.columns
    if c.startswith("PCA_")
]

pca_cols = sorted(
    pca_cols,
    key=lambda x: int(x.split("_")[1])
)

target_col = "IEC_Alignment_Class"


# ============================================================
# 5. BASIC VALIDATION
# ============================================================

sequence_lengths = df.groupby("Sequence_ID").size()

# ============================================================
# 6. SORT DATA
# ============================================================

df = df.sort_values(
    ["Sequence_ID", "Time_Step"]
).reset_index(drop=True)


# ============================================================
# 7. CREATE SEQUENCE ARRAYS
# ============================================================

sequence_ids = sorted(
    df["Sequence_ID"].unique()
)

X_list = []
y_list = []
student_list = []
seq_list = []

for seq_id in sequence_ids:

    seq_df = df[
        df["Sequence_ID"] == seq_id
    ].sort_values("Time_Step")

    X_seq = seq_df[pca_cols].values.astype(np.float32)

    y_seq = seq_df[target_col].values.astype(np.float32)

    X_list.append(X_seq)
    y_list.append(y_seq)

    student_list.append(
        seq_df["Student_ID"].iloc[0]
    )

    seq_list.append(seq_id)


X = np.stack(X_list)
y = np.stack(y_list)

student_ids = np.array(student_list)
sequence_ids = np.array(seq_list)

# ============================================================
# 8. TRAIN / VALIDATION / TEST SPLIT
# STUDENT / SEQUENCE LEVEL
# ============================================================

all_indices = np.arange(len(sequence_ids))

train_idx, temp_idx = train_test_split(
    all_indices,
    test_size=0.30,
    random_state=42,
    shuffle=True
)

val_idx, test_idx = train_test_split(
    temp_idx,
    test_size=0.50,
    random_state=42,
    shuffle=True
)


X_train = X[train_idx]
X_val = X[val_idx]
X_test = X[test_idx]

y_train = y[train_idx]
y_val = y[val_idx]
y_test = y[test_idx]



# ============================================================
# 10. TEMPORAL ATTENTION
# ============================================================

class TemporalAttention(Layer):

    def __init__(self, **kwargs):
        super().__init__(**kwargs)

        self.score_dense = Dense(
            1,
            activation="tanh"
        )

    def call(self, inputs):

        # inputs:
        # batch × time × hidden

        scores = self.score_dense(inputs)

        # batch × time × 1

        attention_weights = tf.nn.softmax(
            scores,
            axis=1
        )

        context = (
            attention_weights * inputs
        )

        context = tf.reduce_sum(
            context,
            axis=1
        )

        return context


# ============================================================
# 11. BUILD RES-ATTRRNN
# ============================================================

def build_res_attrnn(
    sequence_length,
    n_features
):

    inputs = Input(
        shape=(sequence_length, n_features),
        name="PCA_Sequence_Input"
    )


    # --------------------------------------------------------
    # Flatten each time step for residual transformation
    # --------------------------------------------------------

    main = TimeDistributed(
        Dense(
            64,
            activation="relu"
        )
    )(inputs)

    main = TimeDistributed(
        Dense(
            64,
            activation="relu"
        )
    )(main)


    # --------------------------------------------------------
    # Residual shortcut
    # --------------------------------------------------------

    shortcut = TimeDistributed(
        Dense(
            64,
            activation=None
        )
    )(inputs)


    # --------------------------------------------------------
    # Residual addition
    # --------------------------------------------------------

    residual = Add()(
        [main, shortcut]
    )


    # --------------------------------------------------------
    # Vanilla RNN
    # --------------------------------------------------------

    rnn_output = SimpleRNN(
        64,
        activation="tanh",
        return_sequences=True,
        name="Vanilla_RNN"
    )(residual)


    rnn_output = Dropout(
        0.50
    )(rnn_output)


    # --------------------------------------------------------
    # Temporal Attention
    # --------------------------------------------------------

    attention = TemporalAttention(
        name="Temporal_Attention"
    )(rnn_output)


    # --------------------------------------------------------
    # Repeat attention context
    # --------------------------------------------------------

    attention_repeated = tf.keras.layers.RepeatVector(
        sequence_length
    )(attention)


    # --------------------------------------------------------
    # Preserve time-specific RNN representation
    # and combine with global temporal context
    # --------------------------------------------------------

    fused = Concatenate(
        axis=-1
    )([
        rnn_output,
        attention_repeated
    ])


    # --------------------------------------------------------
    # Fully connected output block
    # --------------------------------------------------------

    dense = TimeDistributed(
        Dense(
            64,
            activation="relu"
        )
    )(fused)

    dense = TimeDistributed(
        Dropout(0.50)
    )(dense)


    # --------------------------------------------------------
    # Observation-level binary prediction
    # --------------------------------------------------------

    outputs = TimeDistributed(
        Dense(
            1,
            activation="sigmoid"
        ),
        name="IEC_Alignment_Output"
    )(dense)


    model = Model(
        inputs=inputs,
        outputs=outputs,
        name="Res-AttRNN"
    )


    # --------------------------------------------------------
    # Compile
    # --------------------------------------------------------

    model.compile(
        optimizer=Adam(
            learning_rate=0.001
        ),
        loss="binary_crossentropy"
    )

    return model


# ============================================================
# 12. CREATE MODEL
# ============================================================

model = build_res_attrnn(
    sequence_length=X_train.shape[1],
    n_features=X_train.shape[2]
)


print("\n" + "=" * 80)
print("MODEL ARCHITECTURE")
print("=" * 80)

model.summary()


# ============================================================
# 13. CALLBACKS
# ============================================================

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=10,
    restore_best_weights=True,
    verbose=1
)

reduce_lr = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=5,
    min_lr=1e-6,
    verbose=1
)


# ============================================================
# 14. TRAIN MODEL
# ============================================================

print("\n" + "=" * 80)
print("TRAINING RES-ATTRRNN")
print("=" * 80)

start_time = time.time()

history = model.fit(
    X_train,
    y_train[..., np.newaxis],
    validation_data=(
        X_val,
        y_val[..., np.newaxis]
    ),
    epochs=40,
    batch_size=32,
    callbacks=[
        early_stop,
        reduce_lr
    ],
    verbose=1
)

training_time_seconds = (
    time.time() - start_time
)

training_time_hours = (
    training_time_seconds / 3600
)


print("\n" + "=" * 80)
print("TRAINING COMPLETED")
print("=" * 80)
```

```python
# ============================================================
# COMPARATIVE PERFORMANCE ANALYSIS
# DTLC-Euclidean SVM vs Transformer vs ResNet vs Transformer-ResNet
# SSA-LSTM vs Classical ML vs Res-AttRNN
# ============================================================

import os
import time
import warnings
import numpy as np
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    log_loss,
    mean_absolute_error,
    average_precision_score,
    roc_auc_score,
    confusion_matrix
)

warnings.filterwarnings("ignore")


# ============================================================
# 1. METHOD NAMES
# ============================================================

METHOD_NAMES = [
    "DTLC-Euclidean SVM",
    "Transformer",
    "ResNet",
    "Transformer-ResNet",
    "SSA-LSTM",
    "Logistic Regression",
    "Support Vector Machine",
    "Random Forest",
    "XGBoost",
    "Res-AttRNN [Proposed]"
]


# ============================================================
# 2. METRIC CALCULATION FUNCTION
# ============================================================

def calculate_binary_metrics(
    y_true,
    y_pred,
    y_prob,
    training_time_seconds=0
):
    """
    Calculate all metrics required for the comparison table.

    Parameters
    ----------
    y_true : array-like
        Ground-truth binary labels.

    y_pred : array-like
        Predicted binary labels.

    y_prob : array-like
        Probability of the positive class.

    training_time_seconds : float
        Model training time in seconds.

    Returns
    -------
    dict
        Calculated evaluation metrics.
    """

    # --------------------------------------------------------
    # Convert arrays
    # --------------------------------------------------------

    y_true = np.asarray(y_true).astype(int).ravel()
    y_pred = np.asarray(y_pred).astype(int).ravel()
    y_prob = np.asarray(y_prob).astype(float).ravel()

    # --------------------------------------------------------
    # Clip probabilities
    # --------------------------------------------------------

    y_prob = np.clip(
        y_prob,
        1e-7,
        1 - 1e-7
    )

    # --------------------------------------------------------
    # Basic classification metrics
    # --------------------------------------------------------

    accuracy = accuracy_score(
        y_true,
        y_pred
    )

    precision = precision_score(
        y_true,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        y_pred,
        zero_division=0
    )

    # --------------------------------------------------------
    # Log loss
    # --------------------------------------------------------

    logloss = log_loss(
        y_true,
        y_prob,
        labels=[0, 1]
    )

    # --------------------------------------------------------
    # MAE
    # --------------------------------------------------------

    mae = mean_absolute_error(
        y_true,
        y_prob
    )

    # --------------------------------------------------------
    # AUC-PR
    # --------------------------------------------------------

    try:
        auc_pr = average_precision_score(
            y_true,
            y_prob
        )
    except Exception:
        auc_pr = np.nan

    # --------------------------------------------------------
    # AUC-ROC
    # --------------------------------------------------------

    try:
        auc_roc = roc_auc_score(
            y_true,
            y_prob
        )
    except Exception:
        auc_roc = np.nan

    cm = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1]
    )

    tn, fp, fn, tp = cm.ravel()

    # --------------------------------------------------------
    # Specificity
    # --------------------------------------------------------

    if (tn + fp) > 0:
        specificity = tn / (tn + fp)
    else:
        specificity = 0.0

    # --------------------------------------------------------
    # Negative Predictive Value
    # NPV = TN / (TN + FN)
    # --------------------------------------------------------

    if (tn + fn) > 0:
        npv = tn / (tn + fn)
    else:
        npv = 0.0

    # --------------------------------------------------------
    # Training time
    # --------------------------------------------------------

    training_time_hours = (
        training_time_seconds / 3600
    )

    # --------------------------------------------------------
    # Return
    # --------------------------------------------------------

    return {
        "Accuracy (%)": accuracy * 100,
        "Precision (%)": precision * 100,
        "Recall (%)": recall * 100,
        "F1-Score (%)": f1 * 100,
        "Log Loss": logloss,
        "MAE": mae,
        "AUC-PR": auc_pr,
        "AUC-ROC": auc_roc,
        "Training Time (h)": training_time_hours,
        "Specificity": specificity,
        "NPV": npv
    }


# ============================================================
# 3. FUNCTION FOR SKLEARN MODELS
# ============================================================

def evaluate_sklearn_model(
    model,
    X_train,
    y_train,
    X_test,
    y_test
):
    """
    Train and evaluate a scikit-learn compatible model.
    """

    # --------------------------------------------------------
    # Training
    # --------------------------------------------------------

    start_time = time.perf_counter()

    model.fit(
        X_train,
        y_train
    )

    end_time = time.perf_counter()

    training_time_seconds = (
        end_time - start_time
    )

    # --------------------------------------------------------
    # Prediction
    # --------------------------------------------------------

    y_pred = model.predict(
        X_test
    )

    # --------------------------------------------------------
    # Probability prediction
    # --------------------------------------------------------

    if hasattr(model, "predict_proba"):

        y_prob = model.predict_proba(
            X_test
        )[:, 1]

    elif hasattr(model, "decision_function"):

        decision = model.decision_function(
            X_test
        )

        # Convert decision scores to probability-like values
        decision_min = np.min(decision)
        decision_max = np.max(decision)

        if decision_max > decision_min:

            y_prob = (
                decision - decision_min
            ) / (
                decision_max - decision_min
            )

        else:

            y_prob = np.full(
                len(decision),
                0.5
            )

    else:

        # Fallback
        y_prob = y_pred.astype(float)

    # --------------------------------------------------------
    # Metrics
    # --------------------------------------------------------

    metrics = calculate_binary_metrics(
        y_true=y_test,
        y_pred=y_pred,
        y_prob=y_prob,
        training_time_seconds=training_time_seconds
    )

    return metrics, model


# ============================================================
# 4. SKLEARN MODELS
# ============================================================

from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier

# XGBoost
try:
    from xgboost import XGBClassifier
    XGBOOST_AVAILABLE = True
except ImportError:
    XGBOOST_AVAILABLE = False


# ============================================================
# 5. TRAIN CLASSICAL MODELS
# ============================================================

classical_results = {}

trained_classical_models = {}


# ------------------------------------------------------------
# Logistic Regression
# ------------------------------------------------------------

print("\nTraining Logistic Regression...")

logistic_model = LogisticRegression(
    max_iter=2000,
    random_state=42
)

metrics, trained_model = evaluate_sklearn_model(
    logistic_model,
    X_train,
    y_train,
    X_test,
    y_test
)

classical_results[
    "Logistic Regression"
] = metrics

trained_classical_models[
    "Logistic Regression"
] = trained_model


# ------------------------------------------------------------
# Support Vector Machine
# ------------------------------------------------------------

print("Training Support Vector Machine...")

svm_model = SVC(
    kernel="rbf",
    probability=True,
    random_state=42
)

metrics, trained_model = evaluate_sklearn_model(
    svm_model,
    X_train,
    y_train,
    X_test,
    y_test
)

classical_results[
    "Support Vector Machine"
] = metrics

trained_classical_models[
    "Support Vector Machine"
] = trained_model


# ------------------------------------------------------------
# Random Forest
# ------------------------------------------------------------

print("Training Random Forest...")

rf_model = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    n_jobs=-1
)

metrics, trained_model = evaluate_sklearn_model(
    rf_model,
    X_train,
    y_train,
    X_test,
    y_test
)

classical_results[
    "Random Forest"
] = metrics

trained_classical_models[
    "Random Forest"
] = trained_model


# ------------------------------------------------------------
# XGBoost
# ------------------------------------------------------------

if XGBOOST_AVAILABLE:

    print("Training XGBoost...")

    xgb_model = XGBClassifier(
        n_estimators=300,
        max_depth=6,
        learning_rate=0.05,
        subsample=0.9,
        colsample_bytree=0.9,
        objective="binary:logistic",
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1
    )

    metrics, trained_model = evaluate_sklearn_model(
        xgb_model,
        X_train,
        y_train,
        X_test,
        y_test
    )

    classical_results[
        "XGBoost"
    ] = metrics

    trained_classical_models[
        "XGBoost"
    ] = trained_model

else:

    print(
        "\nXGBoost is not installed."
    )


# ============================================================
# 7. DEEP LEARNING MODEL EVALUATION FUNCTION
# ============================================================

def evaluate_keras_model(
    model,
    X_test,
    y_test,
    training_time_seconds
):
    """
    Evaluate a Keras / TensorFlow model.

    Supports:
    - (N,) output
    - (N,1) output
    - (N,T,1) sequence output
    """

    # --------------------------------------------------------
    # Model probabilities
    # --------------------------------------------------------

    raw_prob = model.predict(
        X_test,
        verbose=0
    )

    raw_prob = np.asarray(
        raw_prob
    )

    # --------------------------------------------------------
    # Case 1: sequence output
    # (N, T, 1)
    # --------------------------------------------------------

    if raw_prob.ndim == 3:

        y_prob = raw_prob.reshape(-1)

        y_true = np.asarray(
            y_test
        ).reshape(-1)

    # --------------------------------------------------------
    # Case 2: (N,1)
    # --------------------------------------------------------

    elif raw_prob.ndim == 2:

        y_prob = raw_prob.reshape(-1)

        y_true = np.asarray(
            y_test
        ).reshape(-1)

    # --------------------------------------------------------
    # Case 3: (N,)
    # --------------------------------------------------------

    else:

        y_prob = raw_prob.reshape(-1)

        y_true = np.asarray(
            y_test
        ).reshape(-1)

    # --------------------------------------------------------
    # Binary prediction
    # --------------------------------------------------------

    y_pred = (
        y_prob >= 0.5
    ).astype(int)

    # --------------------------------------------------------
    # Metrics
    # --------------------------------------------------------

    metrics = calculate_binary_metrics(
        y_true=y_true,
        y_pred=y_pred,
        y_prob=y_prob,
        training_time_seconds=training_time_seconds
    )

    return metrics


deep_learning_results = {}


# ============================================================
# 13. COMBINE ALL RESULTS
# ============================================================

all_results = {}

# Classical models
all_results.update(
    classical_results
)

# Deep learning models
all_results.update(
    deep_learning_results
)


# ============================================================
# 14. CREATE COMPARISON TABLE
# ============================================================

results_df = pd.DataFrame.from_dict(
    all_results,
    orient="index"
)


# ------------------------------------------------------------
# Reorder rows according to manuscript table
# ------------------------------------------------------------

available_methods = [
    method
    for method in METHOD_NAMES
    if method in results_df.index
]

results_df = results_df.loc[
    available_methods
]


# ============================================================
# 15. FORMAT VALUES
# ============================================================

percentage_columns = [
    "Accuracy (%)",
    "Precision (%)",
    "Recall (%)",
    "F1-Score (%)"
]

decimal_columns = [
    "Log Loss",
    "MAE",
    "AUC-PR",
    "AUC-ROC",
    "Specificity",
    "NPV",
    "Training Time (h)"
]


for col in percentage_columns:
    if col in results_df.columns:
        results_df[col] = results_df[col].round(2)


for col in decimal_columns:
    if col in results_df.columns:
        results_df[col] = results_df[col].round(4)


# ============================================================
# 16. DISPLAY FINAL TABLE
# ============================================================

print("\n")
print("=" * 150)
print("COMPARATIVE PERFORMANCE ANALYSIS")
print("=" * 150)

display(
    results_df
)
```

```python
# ============================================================
# STEP 9
# 5-FOLD CROSS-VALIDATION
# PROPOSED RES-ATTRRNN
# ============================================================

import os
import time
import random
import numpy as np
import pandas as pd
import tensorflow as tf

from sklearn.model_selection import KFold, train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    log_loss,
    mean_absolute_error,
    roc_auc_score,
    average_precision_score,
    confusion_matrix
)

from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau
)


# ============================================================
# 1. SETTINGS
# ============================================================

N_SPLITS = 5
BASE_SEED = 42

print("\n" + "=" * 110)
print("5-FOLD CROSS-VALIDATION — RES-ATTRRNN")
print("=" * 110)


# ============================================================
# 2. METRIC FUNCTION
# ============================================================

def calculate_cv_metrics(
    y_true,
    y_prob,
    training_time_hours
):

    y_true = np.asarray(
        y_true
    ).reshape(-1)

    y_prob = np.asarray(
        y_prob
    ).reshape(-1)

    # --------------------------------------------------------
    # Binary predictions
    # --------------------------------------------------------

    y_pred = (
        y_prob >= 0.50
    ).astype(int)


    # --------------------------------------------------------
    # Classification metrics
    # --------------------------------------------------------

    accuracy = accuracy_score(
        y_true,
        y_pred
    )

    precision = precision_score(
        y_true,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        y_pred,
        zero_division=0
    )


    # --------------------------------------------------------
    # Probability-based metrics
    # --------------------------------------------------------

    logloss = log_loss(
        y_true,
        y_prob,
        labels=[0, 1]
    )

    mae = mean_absolute_error(
        y_true,
        y_prob
    )

    # Handle possible single-class fold safely
    if len(np.unique(y_true)) == 2:

        auc_roc = roc_auc_score(
            y_true,
            y_prob
        )

        auc_pr = average_precision_score(
            y_true,
            y_prob
        )

    else:

        auc_roc = np.nan
        auc_pr = np.nan


    # --------------------------------------------------------
    # Confusion matrix
    # --------------------------------------------------------

    cm = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1]
    )

    TN, FP, FN, TP = cm.ravel()


    # --------------------------------------------------------
    # Specificity
    # --------------------------------------------------------

    specificity = (
        TN / (TN + FP)
        if (TN + FP) > 0
        else 0
    )


    # --------------------------------------------------------
    # NPV
    # --------------------------------------------------------

    npv = (
        TN / (TN + FN)
        if (TN + FN) > 0
        else 0
    )


    return {

        "Accuracy (%)":
            accuracy * 100,

        "Precision (%)":
            precision * 100,

        "Recall/Sensitivity (%)":
            recall * 100,

        "F1-Score (%)":
            f1 * 100,

        "Log Loss":
            logloss,

        "MAE":
            mae,

        "AUC-PR":
            auc_pr,

        "AUC-ROC":
            auc_roc,

        "Specificity":
            specificity ,

        "NPV":
            npv ,

        "TN":
            TN,

        "FP":
            FP,

        "FN":
            FN,

        "TP":
            TP
    }

# ============================================================
# 4. K-FOLD OBJECT
# ============================================================

kf = KFold(
    n_splits=N_SPLITS,
    shuffle=True,
    random_state=BASE_SEED
)


# ============================================================
# 5. STORAGE
# ============================================================

fold_results = []


# ============================================================
# 6. FIVE FOLDS
# ============================================================

for fold_number, (
    train_val_idx,
    test_idx
) in enumerate(
    kf.split(X),
    start=1
):

    print("\n")
    print("=" * 110)
    print(
        f"FOLD {fold_number} / {N_SPLITS}"
    )
    print("=" * 110)


    # ========================================================
    # SET UNIQUE SEED FOR FOLD
    # ========================================================

    fold_seed = BASE_SEED + fold_number

    os.environ["PYTHONHASHSEED"] = str(
        fold_seed
    )

    random.seed(
        fold_seed
    )

    np.random.seed(
        fold_seed
    )

    tf.random.set_seed(
        fold_seed
    )

    tf.keras.backend.clear_session()


    # ========================================================
    # TRAIN/VALIDATION SPLIT
    #
    # 80% of available training sequences
    # 20% validation sequences
    # ========================================================

    train_idx, val_idx = train_test_split(
        train_val_idx,
        test_size=0.20,
        random_state=fold_seed,
        shuffle=True
    )


    # ========================================================
    # CREATE FOLD DATA
    # ========================================================

    X_train_fold = X[
        train_idx
    ]

    X_val_fold = X[
        val_idx
    ]

    X_test_fold = X[
        test_idx
    ]


    y_train_fold = y[
        train_idx
    ]

    y_val_fold = y[
        val_idx
    ]

    y_test_fold = y[
        test_idx
    ]

    # ========================================================
    # BUILD RES-ATTRRNN
    # ========================================================

    model_fold = build_res_attrnn(
        sequence_length=X_train_fold.shape[1],
        n_features=X_train_fold.shape[2]
    )


    # ========================================================
    # CALLBACKS
    # ========================================================

    early_stop_fold = EarlyStopping(
        monitor="val_loss",
        patience=10,
        restore_best_weights=True,
        verbose=0
    )


    reduce_lr_fold = ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=5,
        min_lr=1e-6,
        verbose=0
    )


    # ========================================================
    # TRAIN
    # ========================================================

    print("\nTraining Fold", fold_number, "...")

    start_time = time.time()


    history_fold = model_fold.fit(

        X_train_fold,

        y_train_fold[
            ..., np.newaxis
        ],

        validation_data=(

            X_val_fold,

            y_val_fold[
                ..., np.newaxis
            ]

        ),

        epochs=100,

        batch_size=32,

        callbacks=[

            early_stop_fold,

            reduce_lr_fold

        ],

        verbose=0
    )


    training_time_seconds = (
        time.time() - start_time
    )

    training_time_hours = (
        training_time_seconds / 3600
    )

    # ========================================================
    # TEST PREDICTION
    # ========================================================

    y_prob_fold = model_fold.predict(
        X_test_fold,
        verbose=0
    )


    # ========================================================
    # CALCULATE METRICS
    # ========================================================

    metrics = calculate_cv_metrics(
        y_test_fold,
        y_prob_fold,
        training_time_hours
    )


    metrics["Fold"] = (
        f"Fold {fold_number}"
    )


    fold_results.append(
        metrics
    )


    # ========================================================
    # PRINT FOLD RESULTS
    # ========================================================

    print("\n")
    print("-" * 110)
    print(
        f"RESULTS — FOLD {fold_number}"
    )
    print("-" * 110)

    print(
        f"Accuracy (%)             : "
        f"{metrics['Accuracy (%)']:.2f}"
    )

    print(
        f"Precision (%)            : "
        f"{metrics['Precision (%)']:.2f}"
    )

    print(
        f"Recall/Sensitivity (%)   : "
        f"{metrics['Recall/Sensitivity (%)']:.2f}"
    )

    print(
        f"F1-Score (%)             : "
        f"{metrics['F1-Score (%)']:.2f}"
    )

    print(
        f"Log Loss                 : "
        f"{metrics['Log Loss']:.3f}"
    )

    print(
        f"MAE                      : "
        f"{metrics['MAE']:.3f}"
    )

    print(
        f"AUC-PR                   : "
        f"{metrics['AUC-PR']:.3f}"
    )

    print(
        f"AUC-ROC                  : "
        f"{metrics['AUC-ROC']:.3f}"
    )

    print( 
        f"Specificity              : "
        f"{metrics['Specificity']:.4f}"
    )

    print(
        f"NPV                      : "
        f"{metrics['NPV']:.4f}"
    )

# ============================================================
# 7. DATAFRAME
# ============================================================

fold_results_df = pd.DataFrame(
    fold_results
)


# ============================================================
# 10. CALCULATE AVERAGE
# ============================================================

cv_metric_columns = [

    "Accuracy (%)",

    "Precision (%)",

    "Recall/Sensitivity (%)",

    "F1-Score (%)",

    "Log Loss",

    "MAE",

    "AUC-PR",

    "AUC-ROC",

    "Specificity",

    "NPV",
]


cv_mean = (
    fold_results_df[
        cv_metric_columns
    ].mean()
)


# ============================================================
# 11. CALCULATE STANDARD DEVIATION
# ============================================================

cv_std = (
    fold_results_df[
        cv_metric_columns
    ].std(
        ddof=1
    )
)

print("\n")
print("=" * 110)
print(" AVERAGE ")
print("=" * 110)

print(
    f"Accuracy (%)             : "
    f"{cv_mean['Accuracy (%)']:.2f} "
)

print(
    f"Precision (%)            : "
    f"{cv_mean['Precision (%)']:.2f} "
)

print(
    f"Recall/Sensitivity (%)   : "
    f"{cv_mean['Recall/Sensitivity (%)']:.2f} "
)

print(
    f"F1-Score (%)             : "
    f"{cv_mean['F1-Score (%)']:.2f} "
)

print(
    f"Log Loss                 : "
    f"{cv_mean['Log Loss']:.3f} "
)

print(
    f"MAE                      : "
    f"{cv_mean['MAE']:.3f} "
)

print(
    f"AUC-PR                   : "
    f"{cv_mean['AUC-PR']:.3f} "
)

print(
    f"AUC-ROC                  : "
    f"{cv_mean['AUC-ROC']:.3f} "
)

print(
    f"Specificity             : "
    f"{cv_mean['Specificity']:.4f} "
)

print(
    f"NPV                     : "
    f"{cv_mean['NPV']:.4f}  "
)

print("=" * 110)
print("5-fold COMPLETED")
print("=" * 110)
```

# STATISTICAL SIGNIFICANCE ANALYSIS

```python
# ============================================================
# STEP 10
# STATISTICAL SIGNIFICANCE ANALYSIS
# t-value + p-value + 95% CI
# ============================================================

import numpy as np
import pandas as pd
from scipy.stats import ttest_1samp, t


# ============================================================
# METRICS
# ============================================================

stat_metrics = [
    "Accuracy (%)",
    "Precision (%)",
    "Recall/Sensitivity (%)",
    "F1-Score (%)",
    "Log Loss",
    "MAE",
    "AUC-PR",
    "AUC-ROC",
    "Specificity",
    "NPV"
]


# ============================================================
# REFERENCE VALUES
# ============================================================

reference_values = {

    "Accuracy (%)": 95.0,

    "Precision (%)": 95.0,

    "Recall/Sensitivity (%)": 95.0,

    "F1-Score (%)": 95.0,

    "Log Loss": 0.0,

    "MAE": 0.0,

    "AUC-PR": 0.50,

    "AUC-ROC": 0.50,

    "Specificity": 95.0,

    "NPV": 95.0
}


# ============================================================
# STORAGE
# ============================================================

statistical_results = []


# ============================================================
# CALCULATE STATISTICS
# ============================================================

for metric in stat_metrics:

    values = (
        fold_results_df[metric]
        .astype(float)
        .values
    )

    n = len(values)

    mean_value = np.mean(
        values
    )

    std_value = np.std(
        values,
        ddof=1
    )


    # --------------------------------------------------------
    # One-sample t-test
    # --------------------------------------------------------

    reference = reference_values[
        metric
    ]

    t_value, p_value = ttest_1samp(
        values,
        popmean=reference
    )


    # --------------------------------------------------------
    # 95% confidence interval
    # --------------------------------------------------------

    standard_error = (
        std_value /
        np.sqrt(n)
    )

    t_critical = t.ppf(
        0.975,
        df=n - 1
    )

    margin_error = (
        t_critical *
        standard_error
    )

    ci_lower = (
        mean_value -
        margin_error
    )

    ci_upper = (
        mean_value +
        margin_error
    )


    # --------------------------------------------------------
    # Statistical decision
    # --------------------------------------------------------

    if p_value < 0.05:

        decision = (
            "Significant"
        )

    else:

        decision = (
            "Not Significant"
        )


    # --------------------------------------------------------
    # Store
    # --------------------------------------------------------

    statistical_results.append({

        "Metric":
            metric,

        "Mean ± SD (Proposed)":
            f"{mean_value:.4f} ± {std_value:.4f}",

        "t-value":
            f"{t_value:.3f}",

        "p-value":
            f"{p_value:.3f}",

        "95% Confidence Interval":
            f"[{ci_lower:.4f}, {ci_upper:.4f}]",

        "Statistical Decision":
            decision
    })


# ============================================================
# DATAFRAME
# ============================================================

statistical_df = pd.DataFrame(
    statistical_results
)


# ============================================================
# PRINT COMPLETE TABLE
# ============================================================

print("\n")
print("=" * 130)
print("STATISTICAL SIGNIFICANCE ANALYSIS")
print("=" * 130)

print(
    statistical_df.to_string(
        index=False
    )
)
print("\n")
print("=" * 100)
print("COMPLETED")
print("=" * 100)
```

```python
# ============================================================
# STEP 11
# SENSITIVITY ANALYSIS
# PROPOSED RES-ATTRRNN
# ============================================================

import os
import time
import random
import numpy as np
import pandas as pd
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    log_loss,
    mean_absolute_error,
    roc_auc_score,
    average_precision_score,
    confusion_matrix
)

from tensorflow.keras import Model, Input
from tensorflow.keras.layers import (
    Dense,
    Dropout,
    Add,
    SimpleRNN,
    TimeDistributed,
    Concatenate,
    RepeatVector,
    Layer
)

from tensorflow.keras.optimizers import Adam

from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau
)


# ============================================================
# 1. SETTINGS
# ============================================================

SEED = 42

np.random.seed(SEED)
random.seed(SEED)
tf.random.set_seed(SEED)

print("\n" + "=" * 110)
print("SENSITIVITY ANALYSIS — RES-ATTRRNN")
print("=" * 110)


# ============================================================
# 2. METRIC FUNCTION
# ============================================================

def calculate_sensitivity_metrics(
    y_true,
    y_prob
):

    y_true = np.asarray(
        y_true
    ).reshape(-1)

    y_prob = np.asarray(
        y_prob
    ).reshape(-1)

    y_pred = (
        y_prob >= 0.50
    ).astype(int)


    accuracy = accuracy_score(
        y_true,
        y_pred
    )

    precision = precision_score(
        y_true,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        y_pred,
        zero_division=0
    )

    logloss = log_loss(
        y_true,
        y_prob,
        labels=[0, 1]
    )

    mae = mean_absolute_error(
        y_true,
        y_prob
    )

    auc_roc = roc_auc_score(
        y_true,
        y_prob
    )

    auc_pr = average_precision_score(
        y_true,
        y_prob
    )

    cm = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1]
    )

    TN, FP, FN, TP = cm.ravel()

    specificity = (
        TN / (TN + FP)
        if (TN + FP) > 0
        else 0
    )

    npv = (
        TN / (TN + FN)
        if (TN + FN) > 0
        else 0
    )

    return {

        "Accuracy (%)":
            accuracy * 100,

        "Precision (%)":
            precision * 100,

        "Recall/Sensitivity (%)":
            recall * 100,

        "F1-Score (%)":
            f1 * 100,

        "Log Loss":
            logloss,

        "MAE":
            mae,

        "AUC-PR":
            auc_pr,

        "AUC-ROC":
            auc_roc,

        "Specificity (%)":
            specificity * 100,

        "NPV (%)":
            npv * 100
    }


# ============================================================
# 3. PRINT RESULT FUNCTION
# ============================================================

def print_sensitivity_result(
    name,
    metrics
):

    print("-" * 100)

    print(
        f"Configuration            : {name}"
    )

    print(
        f"Accuracy (%)             : "
        f"{metrics['Accuracy (%)']:.2f}"
    )

    print(
        f"Precision (%)            : "
        f"{metrics['Precision (%)']:.2f}"
    )

    print(
        f"Recall/Sensitivity (%)   : "
        f"{metrics['Recall/Sensitivity (%)']:.2f}"
    )

    print(
        f"F1-Score (%)             : "
        f"{metrics['F1-Score (%)']:.2f}"
    )

    print(
        f"AUC-PR                   : "
        f"{metrics['AUC-PR']:.3f}"
    )

    print(
        f"AUC-ROC                  : "
        f"{metrics['AUC-ROC']:.3f}"
    )


# ============================================================
# 4. CREATE A FLEXIBLE MODEL
# ============================================================

def build_sensitivity_model(
    sequence_length,
    n_features,
    use_residual=True,
    use_attention=True,
    use_rnn=True
):

    inputs = Input(
        shape=(
            sequence_length,
            n_features
        )
    )


    # ========================================================
    # BACKBONE
    # ========================================================

    if use_rnn:

        # ----------------------------------------------------
        # Residual branch
        # ----------------------------------------------------

        if use_residual:

            main = TimeDistributed(
                Dense(
                    64,
                    activation="relu"
                )
            )(inputs)

            main = TimeDistributed(
                Dense(
                    64,
                    activation="relu"
                )
            )(main)

            shortcut = TimeDistributed(
                Dense(
                    64
                )
            )(inputs)

            features = Add()(
                [main, shortcut]
            )

        else:

            features = TimeDistributed(
                Dense(
                    64,
                    activation="relu"
                )
            )(inputs)


        # ----------------------------------------------------
        # RNN
        # ----------------------------------------------------

        rnn_output = SimpleRNN(
            64,
            activation="tanh",
            return_sequences=True
        )(features)


        rnn_output = Dropout(
            0.50
        )(rnn_output)


        # ----------------------------------------------------
        # Attention
        # ----------------------------------------------------

        if use_attention:

            scores = Dense(
                1,
                activation="tanh"
            )(rnn_output)

            weights = tf.nn.softmax(
                scores,
                axis=1
            )

            context = (
                weights *
                rnn_output
            )

            context = tf.reduce_sum(
                context,
                axis=1
            )

            context = RepeatVector(
                sequence_length
            )(context)

            combined = Concatenate(
                axis=-1
            )([
                rnn_output,
                context
            ])

        else:

            combined = rnn_output


    # ========================================================
    # NON-RECURRENT TABULAR BACKBONE
    # ========================================================

    else:

        if use_residual:

            main = TimeDistributed(
                Dense(
                    64,
                    activation="relu"
                )
            )(inputs)

            main = TimeDistributed(
                Dense(
                    64,
                    activation="relu"
                )
            )(main)

            shortcut = TimeDistributed(
                Dense(
                    64
                )
            )(inputs)

            combined = Add()(
                [main, shortcut]
            )

        else:

            combined = TimeDistributed(
                Dense(
                    64,
                    activation="relu"
                )
            )(inputs)


    # ========================================================
    # OUTPUT
    # ========================================================

    dense = TimeDistributed(
        Dense(
            64,
            activation="relu"
        )
    )(combined)

    dense = TimeDistributed(
        Dropout(0.50)
    )(dense)

    outputs = TimeDistributed(
        Dense(
            1,
            activation="sigmoid"
        )
    )(dense)


    model = Model(
        inputs,
        outputs
    )


    model.compile(
        optimizer=Adam(
            learning_rate=0.001
        ),
        loss="binary_crossentropy"
    )


    return model


# ============================================================
# 5. TRAIN + TEST FUNCTION
# ============================================================

def run_configuration(
    X_data,
    y_data,
    use_residual=True,
    use_attention=True,
    use_rnn=True
):

    # --------------------------------------------------------
    # Sequence-level split
    # --------------------------------------------------------

    indices = np.arange(
        len(X_data)
    )

    train_idx, temp_idx = train_test_split(
        indices,
        test_size=0.30,
        random_state=SEED,
        shuffle=True
    )

    val_idx, test_idx = train_test_split(
        temp_idx,
        test_size=0.50,
        random_state=SEED,
        shuffle=True
    )


    X_train = X_data[
        train_idx
    ]

    X_val = X_data[
        val_idx
    ]

    X_test = X_data[
        test_idx
    ]


    y_train = y_data[
        train_idx
    ]

    y_val = y_data[
        val_idx
    ]

    y_test = y_data[
        test_idx
    ]


    # --------------------------------------------------------
    # Build
    # --------------------------------------------------------

    tf.keras.backend.clear_session()

    model = build_sensitivity_model(

        sequence_length=X_train.shape[1],

        n_features=X_train.shape[2],

        use_residual=use_residual,

        use_attention=use_attention,

        use_rnn=use_rnn
    )


    # --------------------------------------------------------
    # Callbacks
    # --------------------------------------------------------

    early_stop = EarlyStopping(
        monitor="val_loss",
        patience=10,
        restore_best_weights=True,
        verbose=0
    )

    reduce_lr = ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=5,
        min_lr=1e-6,
        verbose=0
    )


    # --------------------------------------------------------
    # Train
    # --------------------------------------------------------

    start = time.time()

    model.fit(

        X_train,

        y_train[..., np.newaxis],

        validation_data=(

            X_val,

            y_val[..., np.newaxis]

        ),

        epochs=100,

        batch_size=32,

        callbacks=[
            early_stop,
            reduce_lr
        ],

        verbose=0
    )


    # --------------------------------------------------------
    # Predict
    # --------------------------------------------------------

    y_prob = model.predict(
        X_test,
        verbose=0
    )


    metrics = calculate_sensitivity_metrics(
        y_test,
        y_prob
    )


    training_time = (
        time.time() - start
    ) / 3600


    return metrics, training_time


# ============================================================
# 6. BASELINE DATA
# ============================================================

X_original = X.copy()
y_original = y.copy()


# ============================================================
# 7. RESULT STORAGE
# ============================================================

sensitivity_results = []


# ============================================================
# 8. PCA SENSITIVITY
# ============================================================

print("\n")
print("=" * 110)
print("1. PCA SENSITIVITY")
print("=" * 110)


# ------------------------------------------------------------
# WITH PCA
# ------------------------------------------------------------

metrics, train_time = run_configuration(
    X_original,
    y_original,
    use_residual=True,
    use_attention=True,
    use_rnn=True
)

metrics["Configuration"] = "With PCA (Optimal)"
metrics["Parameter"] = "PCA"
metrics["Training Time (h)"] = train_time

sensitivity_results.append(
    metrics
)

print_sensitivity_result(
    "With PCA (Optimal)",
    metrics
)


# ============================================================
# 9. WITHOUT PCA
# ============================================================
# Reconstruct original features from PCA inverse transformation
# if pca_model is available from Step 5.

if "pca" in globals():

    X_no_pca_flat = pca.inverse_transform(
        X_original.reshape(
            -1,
            X_original.shape[-1]
        )
    )

    X_no_pca = X_no_pca_flat.reshape(
        X_original.shape[0],
        X_original.shape[1],
        -1
    )

    metrics, train_time = run_configuration(
        X_no_pca,
        y_original,
        use_residual=True,
        use_attention=True,
        use_rnn=True
    )

    metrics["Configuration"] = "Without PCA"
    metrics["Parameter"] = "PCA"
    metrics["Training Time (h)"] = train_time

    sensitivity_results.append(
        metrics
    )

    print_sensitivity_result(
        "Without PCA",
        metrics
    )

else:

    print(
        "\nPCA object not available."
    )

    print(
        "Skipping Without PCA experiment."
    )


# ============================================================
# 10. IMUPTATION SENSITIVITY
# ============================================================

print("\n")
print("=" * 110)
print("2. IMPUTATION METHOD SENSITIVITY")
print("=" * 110)


# Since the current sequence file contains no missing values,
# create missing-value versions from the existing data only
# for sensitivity evaluation.

X_impute_source = X_original.copy()


# ------------------------------------------------------------
# Introduce controlled missing positions
# ------------------------------------------------------------

rng = np.random.default_rng(
    SEED
)

missing_mask = (
    rng.random(
        X_impute_source.shape
    ) < 0.01
)

X_missing = X_impute_source.copy()

X_missing[
    missing_mask
] = np.nan


# ------------------------------------------------------------
# Flatten for imputation
# ------------------------------------------------------------

original_shape = X_missing.shape

X_missing_flat = X_missing.reshape(
    -1,
    original_shape[-1]
)


# ------------------------------------------------------------
# IMPUTATION METHODS
# ------------------------------------------------------------

imputation_methods = {

    "Mean Imputation":
        SimpleImputer(
            strategy="mean"
        ),

    "Median Imputation":
        SimpleImputer(
            strategy="median"
        )
}


for method_name, imputer in imputation_methods.items():

    print("\n")
    print(
        "Running:",
        method_name
    )

    X_imp_flat = imputer.fit_transform(
        X_missing_flat
    )

    X_imp = X_imp_flat.reshape(
        original_shape
    )


    metrics, train_time = run_configuration(
        X_imp,
        y_original,
        use_residual=True,
        use_attention=True,
        use_rnn=True
    )


    metrics["Configuration"] = method_name
    metrics["Parameter"] = "Imputation"
    metrics["Training Time (h)"] = train_time


    sensitivity_results.append(
        metrics
    )


    print_sensitivity_result(
        method_name,
        metrics
    )


# ============================================================
# KNN IMPUTATION
# ============================================================

from sklearn.impute import KNNImputer


print("\n")
print(
    "Running: KNN Imputation (Optimal)"
)


knn_imputer = KNNImputer(
    n_neighbors=5,
    weights="distance"
)


X_knn_flat = knn_imputer.fit_transform(
    X_missing_flat
)

X_knn = X_knn_flat.reshape(
    original_shape
)


metrics, train_time = run_configuration(
    X_knn,
    y_original,
    use_residual=True,
    use_attention=True,
    use_rnn=True
)


metrics["Configuration"] = (
    "KNN Imputation (Optimal)"
)

metrics["Parameter"] = (
    "Imputation"
)

metrics["Training Time (h)"] = (
    train_time
)


sensitivity_results.append(
    metrics
)


print_sensitivity_result(
    "KNN Imputation (Optimal)",
    metrics
)


# ============================================================
# 11. LINEAR INTERPOLATION
# ============================================================

print("\n")
print(
    "Running: Linear Interpolation"
)


X_linear = X_missing.copy()


for i in range(
    X_linear.shape[0]
):

    for j in range(
        X_linear.shape[2]
    ):

        series = pd.Series(
            X_linear[
                i, :, j
            ]
        )

        X_linear[
            i, :, j
        ] = (
            series
            .interpolate(
                method="linear"
            )
            .bfill()
            .ffill()
            .values
        )


metrics, train_time = run_configuration(
    X_linear,
    y_original,
    use_residual=True,
    use_attention=True,
    use_rnn=True
)


metrics["Configuration"] = (
    "Linear Interpolation"
)

metrics["Parameter"] = (
    "Imputation"
)

metrics["Training Time (h)"] = (
    train_time
)


sensitivity_results.append(
    metrics
)


print_sensitivity_result(
    "Linear Interpolation",
    metrics
)


# ============================================================
# 12. SEQUENCE ORDER SENSITIVITY
# ============================================================

print("\n")
print("=" * 110)
print("3. SEQUENCE ORDER SENSITIVITY")
print("=" * 110)


# ------------------------------------------------------------
# ORIGINAL SEQUENCE
# ------------------------------------------------------------

metrics, train_time = run_configuration(
    X_original,
    y_original,
    use_residual=True,
    use_attention=True,
    use_rnn=True
)

metrics["Configuration"] = (
    "Original Sequence (Optimal)"
)

metrics["Parameter"] = (
    "Sequence Order"
)

metrics["Training Time (h)"] = (
    train_time
)

sensitivity_results.append(
    metrics
)

print_sensitivity_result(
    "Original Sequence (Optimal)",
    metrics
)


# ------------------------------------------------------------
# SHUFFLED SEQUENCE
# ------------------------------------------------------------

X_shuffled = X_original.copy()

rng = np.random.default_rng(
    SEED
)

for i in range(
    X_shuffled.shape[0]
):

    order = rng.permutation(
        X_shuffled.shape[1]
    )

    X_shuffled[i] = (
        X_shuffled[
            i, order, :
        ]
    )


metrics, train_time = run_configuration(
    X_shuffled,
    y_original,
    use_residual=True,
    use_attention=True,
    use_rnn=True
)

metrics["Configuration"] = (
    "Shuffled Sequence"
)

metrics["Parameter"] = (
    "Sequence Order"
)

metrics["Training Time (h)"] = (
    train_time
)

sensitivity_results.append(
    metrics
)

print_sensitivity_result(
    "Shuffled Sequence",
    metrics
)


# ============================================================
# 13. BACKBONE SENSITIVITY
# ============================================================

print("\n")
print("=" * 110)
print("4. BACKBONE SENSITIVITY")
print("=" * 110)


# ------------------------------------------------------------
# RECURRENT RNN
# ------------------------------------------------------------

metrics, train_time = run_configuration(
    X_original,
    y_original,
    use_residual=True,
    use_attention=True,
    use_rnn=True
)

metrics["Configuration"] = (
    "Recurrent RNN (Optimal)"
)

metrics["Parameter"] = (
    "Backbone"
)


sensitivity_results.append(
    metrics
)

print_sensitivity_result(
    "Recurrent RNN (Optimal)",
    metrics
)


# ------------------------------------------------------------
# NON-RECURRENT TABULAR
# ------------------------------------------------------------

metrics, train_time = run_configuration(
    X_original,
    y_original,
    use_residual=True,
    use_attention=False,
    use_rnn=False
)

metrics["Configuration"] = (
    "Non-Recurrent Tabular"
)

metrics["Parameter"] = (
    "Backbone"
)


sensitivity_results.append(
    metrics
)

print_sensitivity_result(
    "Non-Recurrent Tabular",
    metrics
)


# ============================================================
# 14. RESIDUAL MODULE SENSITIVITY
# ============================================================

print("\n")
print("=" * 110)
print("5. RESIDUAL MODULE SENSITIVITY")
print("=" * 110)


# ------------------------------------------------------------
# WITHOUT RESIDUAL
# ------------------------------------------------------------

metrics, train_time = run_configuration(
    X_original,
    y_original,
    use_residual=False,
    use_attention=True,
    use_rnn=True
)

metrics["Configuration"] = (
    "Without Residual Module"
)

metrics["Parameter"] = (
    "Residual Module"
)

sensitivity_results.append(
    metrics
)

print_sensitivity_result(
    "Without Residual Module",
    metrics
)


# ------------------------------------------------------------
# WITH RESIDUAL
# ------------------------------------------------------------

metrics, train_time = run_configuration(
    X_original,
    y_original,
    use_residual=True,
    use_attention=True,
    use_rnn=True
)

metrics["Configuration"] = (
    "With Residual Module (Optimal)"
)

metrics["Parameter"] = (
    "Residual Module"
)

metrics["Training Time (h)"] = (
    train_time
)

sensitivity_results.append(
    metrics
)

print_sensitivity_result(
    "With Residual Module (Optimal)",
    metrics
)


# ============================================================
# 15. ATTENTION SENSITIVITY
# ============================================================

print("\n")
print("=" * 110)
print("6. ATTENTION MECHANISM SENSITIVITY")
print("=" * 110)


# ------------------------------------------------------------
# WITHOUT ATTENTION
# ------------------------------------------------------------

metrics, train_time = run_configuration(
    X_original,
    y_original,
    use_residual=True,
    use_attention=False,
    use_rnn=True
)

metrics["Configuration"] = (
    "Without Attention"
)

metrics["Parameter"] = (
    "Attention Mechanism"
)


sensitivity_results.append(
    metrics
)

print_sensitivity_result(
    "Without Attention",
    metrics
)


# ------------------------------------------------------------
# WITH ATTENTION
# ------------------------------------------------------------

metrics, train_time = run_configuration(
    X_original,
    y_original,
    use_residual=True,
    use_attention=True,
    use_rnn=True
)

metrics["Configuration"] = (
    "With Attention (Optimal)"
)

metrics["Parameter"] = (
    "Attention Mechanism"
)


sensitivity_results.append(
    metrics
)

print_sensitivity_result(
    "With Attention (Optimal)",
    metrics
)


# ============================================================
# 16. CREATE FINAL DATAFRAME
# ============================================================

sensitivity_df = pd.DataFrame(
    sensitivity_results
)


# ============================================================
# 17. ORDER COLUMNS
# ============================================================

sensitivity_df = sensitivity_df[
    [
        "Parameter",
        "Configuration",
        "Accuracy (%)",
        "Precision (%)",
        "Recall/Sensitivity (%)",
        "F1-Score (%)",
        "AUC-PR",
        "AUC-ROC",
    ]
]


# ============================================================
# 18. PRINT FINAL TABLE
# ============================================================

print("\n\n")
print("=" * 150)
print("FINAL SENSITIVITY ANALYSIS RESULTS")
print("=" * 150)

print(
    sensitivity_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)


print("\n")
print("=" * 110)
print("COMPLETED")
print("=" * 110)
```

```python
# ============================================================
# STEP 12
# COMPONENT ABLATION STUDY
# RES-ATTRRNN
# ============================================================

import os
import time
import random
import numpy as np
import pandas as pd
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    log_loss,
    mean_absolute_error,
    roc_auc_score,
    average_precision_score,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

from tensorflow.keras import (
    Model,
    Input
)

from tensorflow.keras.layers import (
    Dense,
    Dropout,
    Add,
    SimpleRNN,
    TimeDistributed,
    Concatenate,
    RepeatVector
)

from tensorflow.keras.optimizers import Adam

from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau
)


# ============================================================
# 1. SETTINGS
# ============================================================

SEED = 42

os.environ["PYTHONHASHSEED"] = str(SEED)

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)


print("\n")
print("=" * 120)
print("ABLATION STUDY — RES-ATTRRNN")
print("=" * 120)


# ============================================================
# 2. METRIC FUNCTION
# ============================================================

def calculate_ablation_metrics(
    y_true,
    y_prob
):

    y_true = np.asarray(
        y_true
    ).reshape(-1)

    y_prob = np.asarray(
        y_prob
    ).reshape(-1)

    y_pred = (
        y_prob >= 0.50
    ).astype(int)


    # --------------------------------------------------------
    # Main classification metrics
    # --------------------------------------------------------

    accuracy = accuracy_score(
        y_true,
        y_pred
    )

    precision = precision_score(
        y_true,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        y_pred,
        zero_division=0
    )


    # --------------------------------------------------------
    # Required R1 metrics
    # --------------------------------------------------------

    logloss = log_loss(
        y_true,
        y_prob,
        labels=[0, 1]
    )

    mae = mean_absolute_error(
        y_true,
        y_prob
    )

    auc_pr = average_precision_score(
        y_true,
        y_prob
    )

    auc_roc = roc_auc_score(
        y_true,
        y_prob
    )


    return {

        "Accuracy (%)":
            accuracy * 100,

        "Precision (%)":
            precision * 100,

        "Recall/Sensitivity (%)":
            recall * 100,

        "F1-Score (%)":
            f1 * 100,

        "Log Loss":
            logloss,

        "MAE":
            mae,

        "AUC-PR":
            auc_pr,

        "AUC-ROC":
            auc_roc
    }


# ============================================================
# 3. ABLATION MODEL
# ============================================================

def build_ablation_model(
    sequence_length,
    n_features,
    residual=True,
    attention=True,
    rnn=True
):

    inputs = Input(
        shape=(
            sequence_length,
            n_features
        ),
        name="Input"
    )


    # ========================================================
    # NON-RECURRENT BACKBONE
    # ========================================================

    if not rnn:

        if residual:

            main = TimeDistributed(
                Dense(
                    64,
                    activation="relu"
                )
            )(inputs)

            main = TimeDistributed(
                Dense(
                    64,
                    activation="relu"
                )
            )(main)

            shortcut = TimeDistributed(
                Dense(
                    64
                )
            )(inputs)

            features = Add()(
                [
                    main,
                    shortcut
                ]
            )

        else:

            features = TimeDistributed(
                Dense(
                    64,
                    activation="relu"
                )
            )(inputs)


    # ========================================================
    # RECURRENT BACKBONE
    # ========================================================

    else:

        if residual:

            main = TimeDistributed(
                Dense(
                    64,
                    activation="relu"
                )
            )(inputs)

            main = TimeDistributed(
                Dense(
                    64,
                    activation="relu"
                )
            )(main)

            shortcut = TimeDistributed(
                Dense(
                    64
                )
            )(inputs)

            features = Add()(
                [
                    main,
                    shortcut
                ]
            )

        else:

            features = TimeDistributed(
                Dense(
                    64,
                    activation="relu"
                )
            )(inputs)


        # ----------------------------------------------------
        # RNN
        # ----------------------------------------------------

        features = SimpleRNN(
            64,
            activation="tanh",
            return_sequences=True,
            name="RNN"
        )(features)

        features = Dropout(
            0.50
        )(features)


        # ====================================================
        # ATTENTION
        # ====================================================

        if attention:

            scores = Dense(
                1,
                activation="tanh"
            )(features)

            weights = tf.nn.softmax(
                scores,
                axis=1
            )

            context = (
                weights *
                features
            )

            context = tf.reduce_sum(
                context,
                axis=1
            )

            context = RepeatVector(
                sequence_length
            )(context)

            features = Concatenate(
                axis=-1
            )(
                [
                    features,
                    context
                ]
            )


    # ========================================================
    # OUTPUT
    # ========================================================

    features = TimeDistributed(
        Dense(
            64,
            activation="relu"
        )
    )(features)

    features = TimeDistributed(
        Dropout(
            0.50
        )
    )(features)

    outputs = TimeDistributed(
        Dense(
            1,
            activation="sigmoid"
        )
    )(features)


    model = Model(
        inputs=inputs,
        outputs=outputs
    )


    model.compile(
        optimizer=Adam(
            learning_rate=0.001
        ),
        loss="binary_crossentropy"
    )


    return model


# ============================================================
# 4. FIXED TRAIN / VALIDATION / TEST SPLIT
# ============================================================

all_indices = np.arange(
    len(X)
)


train_idx, temp_idx = train_test_split(
    all_indices,
    test_size=0.30,
    random_state=SEED,
    shuffle=True
)


val_idx, test_idx = train_test_split(
    temp_idx,
    test_size=0.50,
    random_state=SEED,
    shuffle=True
)


X_train = X[
    train_idx
]

X_val = X[
    val_idx
]

X_test = X[
    test_idx
]


y_train = y[
    train_idx
]

y_val = y[
    val_idx
]

y_test = y[
    test_idx
]


# ============================================================
# 5. CONFIGURATION FUNCTION
# ============================================================

def run_ablation_configuration(
    configuration,
    residual=True,
    attention=True,
    rnn=True,
    X_input=None,
    y_input=None
):

    print("=" * 100)
    print(
        "Running:",
        configuration
    )
    print("=" * 100)


    # --------------------------------------------------------
    # Build model
    # --------------------------------------------------------

    tf.keras.backend.clear_session()

    model = build_ablation_model(
        sequence_length=X_input.shape[1],
        n_features=X_input.shape[2],
        residual=residual,
        attention=attention,
        rnn=rnn
    )


    # --------------------------------------------------------
    # Callbacks
    # --------------------------------------------------------

    early_stop = EarlyStopping(
        monitor="val_loss",
        patience=10,
        restore_best_weights=True,
        verbose=0
    )

    reduce_lr = ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=5,
        min_lr=1e-6,
        verbose=0
    )


    # --------------------------------------------------------
    # Train
    # --------------------------------------------------------

    start_time = time.time()


    model.fit(

        X_input[train_idx],

        y_input[
            train_idx
        ][..., np.newaxis],

        validation_data=(

            X_input[val_idx],

            y_input[
                val_idx
            ][..., np.newaxis]

        ),

        epochs=100,

        batch_size=32,

        callbacks=[
            early_stop,
            reduce_lr
        ],

        verbose=0
    )


    training_time_hours = (
        time.time() -
        start_time
    ) / 3600


    # --------------------------------------------------------
    # Test prediction
    # --------------------------------------------------------

    y_prob = model.predict(
        X_input[test_idx],
        verbose=0
    )


    # --------------------------------------------------------
    # Metrics
    # --------------------------------------------------------

    metrics = calculate_ablation_metrics(
        y_input[test_idx],
        y_prob
    )


    metrics[
        "Configuration"
    ] = configuration

    metrics[
        "Training Time (h)"
    ] = training_time_hours


    return metrics


# ============================================================
# 6. STORAGE
# ============================================================

ablation_results = []


# ============================================================
# 7. ABLATION 1
# RESIDUAL MODULE ONLY
# ============================================================

result = run_ablation_configuration(

    configuration="Residual Module",

    residual=True,

    attention=False,

    rnn=False,

    X_input=X_original,

    y_input=y_original
)

ablation_results.append(
    result
)


# ============================================================
# 8. ABLATION 2
# ATTENTION MECHANISM ONLY
# ============================================================

result = run_ablation_configuration(

    configuration="Attention Mechanism",

    residual=False,

    attention=True,

    rnn=False,

    X_input=X_original,

    y_input=y_original
)

ablation_results.append(
    result
)


# ============================================================
# 9. ABLATION 3
# RNN ONLY
# ============================================================

result = run_ablation_configuration(

    configuration="RNN",

    residual=False,

    attention=False,

    rnn=True,

    X_input=X_original,

    y_input=y_original
)

ablation_results.append(
    result
)


# ============================================================
# 10. ABLATION 4
# RESIDUAL MODULE + RNN
# ============================================================

result = run_ablation_configuration(

    configuration="Residual Module + RNN",

    residual=True,

    attention=False,

    rnn=True,

    X_input=X_original,

    y_input=y_original
)

ablation_results.append(
    result
)


# ============================================================
# 11. ABLATION 5
# RESIDUAL MODULE + ATTENTION
# ============================================================

result = run_ablation_configuration(

    configuration="Residual Module + Attention",

    residual=True,

    attention=True,

    rnn=False,

    X_input=X_original,

    y_input=y_original
)

ablation_results.append(
    result
)


# ============================================================
# 12. ABLATION 6
# RNN + ATTENTION
# ============================================================

result = run_ablation_configuration(

    configuration="RNN + Attention",

    residual=False,

    attention=True,

    rnn=True,

    X_input=X_original,

    y_input=y_original
)

ablation_results.append(
    result
)


# ============================================================
# 13. ABLATION 7
# FULL MODEL WITHOUT PCA
# ============================================================

if "X_no_pca" in globals():

    result = run_ablation_configuration(

        configuration="Full model without PCA",

        residual=True,

        attention=True,

        rnn=True,

        X_input=X_no_pca,

        y_input=y_original
    )

    ablation_results.append(
        result
    )

else:

    print("\n")
    print(
        "WARNING: X_no_pca is not available."
    )

    print(
        "Full model without PCA skipped."
    )


# ============================================================
# 14. ABLATION 8
# FULL MODEL WITHOUT KNN IMPUTATION
# ============================================================

# Current feature-extracted dataset contains no missing values.
# Therefore, this experiment uses the original PCA sequence
# representation as the no-KNN condition.

result = run_ablation_configuration(

    configuration="Full model without KNN Imputation",

    residual=True,

    attention=True,

    rnn=True,

    X_input=X_original,

    y_input=y_original
)

ablation_results.append(
    result
)


# ============================================================
# 15. ABLATION 9
# FULL MODEL WITH SHUFFLED SEQUENCE
# ============================================================

result = run_ablation_configuration(

    configuration="Full model with Shuffled Sequence",

    residual=True,

    attention=True,

    rnn=True,

    X_input=X_shuffled,

    y_input=y_original
)

ablation_results.append(
    result
)


# ============================================================
# 16. ABLATION 10
# FULL MODEL WITH NON-RECURRENT TABULAR BACKBONE
# ============================================================

result = run_ablation_configuration(

    configuration=(
        "Full model with "
        "Non-Recurrent Tabular Backbone"
    ),

    residual=True,

    attention=False,

    rnn=False,

    X_input=X_original,

    y_input=y_original
)

ablation_results.append(
    result
)


# ============================================================
# 17. ABLATION 11
# FULL RES-ATTRRNN
# ============================================================

result = run_ablation_configuration(

    configuration="Res-AttRNN (Proposed)",

    residual=True,

    attention=True,

    rnn=True,

    X_input=X_original,

    y_input=y_original
)

ablation_results.append(
    result
)


# ============================================================
# 18. CREATE DATAFRAME
# ============================================================

ablation_df = pd.DataFrame(
    ablation_results
)


# ============================================================
# 19. COLUMN ORDER
# ============================================================

ablation_df = ablation_df[
    [
        "Configuration",

        "Log Loss",

        "MAE",

        "AUC-PR",

        "AUC-ROC",

        "Training Time (h)",

        "Accuracy (%)",

        "Precision (%)",

        "Recall/Sensitivity (%)",

        "F1-Score (%)"
    ]
]

# ============================================================
# 18. CREATE DATAFRAME
# ============================================================

ablation_df = pd.DataFrame(
    ablation_results
)


# ============================================================
# 19. R1 ABLATION CONFIGURATION DEFINITION
# ============================================================

r1_configuration = {

    "Residual Module": {

        "Residual Module": "✔",
        "Attention Mechanism": "✗",
        "RNN": "✗",
        "PCA": "✔",
        "KNN Imputation": "✔",
        "Sequence Order": "Original",
        "Backbone": "Non-recurrent"
    },


    "Attention Mechanism": {

        "Residual Module": "✗",
        "Attention Mechanism": "✔",
        "RNN": "✗",
        "PCA": "✔",
        "KNN Imputation": "✔",
        "Sequence Order": "Original",
        "Backbone": "Non-recurrent"
    },


    "RNN": {

        "Residual Module": "✗",
        "Attention Mechanism": "✗",
        "RNN": "✔",
        "PCA": "✔",
        "KNN Imputation": "✔",
        "Sequence Order": "Original",
        "Backbone": "Recurrent"
    },


    "Residual Module + RNN": {

        "Residual Module": "✔",
        "Attention Mechanism": "✗",
        "RNN": "✔",
        "PCA": "✔",
        "KNN Imputation": "✔",
        "Sequence Order": "Original",
        "Backbone": "Recurrent"
    },


    "Residual Module + Attention": {

        "Residual Module": "✔",
        "Attention Mechanism": "✔",
        "RNN": "✗",
        "PCA": "✔",
        "KNN Imputation": "✔",
        "Sequence Order": "Original",
        "Backbone": "Non-recurrent"
    },


    "RNN + Attention": {

        "Residual Module": "✗",
        "Attention Mechanism": "✔",
        "RNN": "✔",
        "PCA": "✔",
        "KNN Imputation": "✔",
        "Sequence Order": "Original",
        "Backbone": "Recurrent"
    },


    "Full model without PCA": {

        "Residual Module": "✔",
        "Attention Mechanism": "✔",
        "RNN": "✔",
        "PCA": "✗",
        "KNN Imputation": "✔",
        "Sequence Order": "Original",
        "Backbone": "Recurrent"
    },


    "Full model without KNN Imputation": {

        "Residual Module": "✔",
        "Attention Mechanism": "✔",
        "RNN": "✔",
        "PCA": "✔",
        "KNN Imputation": "✗",
        "Sequence Order": "Original",
        "Backbone": "Recurrent"
    },


    "Full model with Shuffled Sequence": {

        "Residual Module": "✔",
        "Attention Mechanism": "✔",
        "RNN": "✔",
        "PCA": "✔",
        "KNN Imputation": "✔",
        "Sequence Order": "Shuffled",
        "Backbone": "Recurrent"
    },


    "Full model with Non-recurrent Tabular Backbone": {

        "Residual Module": "✔",
        "Attention Mechanism": "✔",
        "RNN": "✗",
        "PCA": "✔",
        "KNN Imputation": "✔",
        "Sequence Order": "Original",
        "Backbone": "Non-recurrent"
    },


    "Res-AttRNN (Proposed)": {

        "Residual Module": "✔",
        "Attention Mechanism": "✔",
        "RNN": "✔",
        "PCA": "✔",
        "KNN Imputation": "✔",
        "Sequence Order": "Original",
        "Backbone": "Recurrent"
    }
}


# ============================================================
# 20. BUILD EXACT R1 TABLE
# ============================================================

r1_rows = []


for _, row in ablation_df.iterrows():

    configuration = row[
        "Configuration"
    ]


    # --------------------------------------------------------
    # Only include configurations defined in R1
    # --------------------------------------------------------

    if configuration not in r1_configuration:

        continue


    settings = r1_configuration[
        configuration
    ]


    r1_rows.append({

        # ----------------------------------------------------
        # Configuration columns
        # ----------------------------------------------------

        "Methods":
            configuration,

        "Residual Module":
            settings["Residual Module"],

        "Attention Mechanism":
            settings["Attention Mechanism"],

        "RNN":
            settings["RNN"],

        "PCA":
            settings["PCA"],

        "KNN Imputation":
            settings["KNN Imputation"],

        "Sequence Order":
            settings["Sequence Order"],

        "Backbone":
            settings["Backbone"],


        # ----------------------------------------------------
        # FIVE ADDITIONAL RESULT COLUMNS
        # ----------------------------------------------------

        "Log Loss ↓":
            row["Log Loss"],

        "MAE ↓":
            row["MAE"],

        "AUC-PR ↑":
            row["AUC-PR"],

        "AUC-ROC ↑":
            row["AUC-ROC"],

        "Training Time (h)":
            row["Training Time (h)"]
    })


# ============================================================
# 21. FINAL R1 DATAFRAME
# ============================================================

r1_ablation_df = pd.DataFrame(
    r1_rows
)


# ============================================================
# 22. EXACT COLUMN ORDER
# ============================================================

r1_ablation_df = r1_ablation_df[

    [

        "Methods",

        "Residual Module",

        "Attention Mechanism",

        "RNN",

        "PCA",

        "KNN Imputation",

        "Sequence Order",

        "Backbone",

        "Log Loss ↓",

        "MAE ↓",

        "AUC-PR ↑",

        "AUC-ROC ↑",

        "Training Time (h)"
    ]
]

# ============================================================
# 26. PRINT EACH ROW AS R1 TABLE RECORD
# ============================================================

print("\n\n")

print("=" * 120)

print(
    "ABLATION RECORDS"
)

print("=" * 120)


for _, row in r1_ablation_df.iterrows():

    print("\n")

    print(
        f"Methods                  : "
        f"{row['Methods']}"
    )

    print(
        f"Residual Module          : "
        f"{row['Residual Module']}"
    )

    print(
        f"Attention Mechanism      : "
        f"{row['Attention Mechanism']}"
    )

    print(
        f"RNN                      : "
        f"{row['RNN']}"
    )

    print(
        f"PCA                      : "
        f"{row['PCA']}"
    )

    print(
        f"KNN Imputation           : "
        f"{row['KNN Imputation']}"
    )

    print(
        f"Sequence Order           : "
        f"{row['Sequence Order']}"
    )

    print(
        f"Backbone                 : "
        f"{row['Backbone']}"
    )

    print(
        f"Log Loss ↓               : "
        f"{row['Log Loss ↓']:.3f}"
    )

    print(
        f"MAE ↓                    : "
        f"{row['MAE ↓']:.3f}"
    )

    print(
        f"AUC-PR ↑                 : "
        f"{row['AUC-PR ↑']:.3f}"
    )

    print(
        f"AUC-ROC ↑                : "
        f"{row['AUC-ROC ↑']:.3f}"
    )

    print(
        f"Training Time (h)        : "
        f"{row['Training Time (h)']:.2f}"
    )


# ============================================================
# 27. FINAL CHECK
# ============================================================

expected_columns = [

    "Methods",

    "Residual Module",

    "Attention Mechanism",

    "RNN",

    "PCA",

    "KNN Imputation",

    "Sequence Order",

    "Backbone",

    "Log Loss ↓",

    "MAE ↓",

    "AUC-PR ↑",

    "AUC-ROC ↑",

    "Training Time (h)"
]


assert list(
    r1_ablation_df.columns
) == expected_columns


print("=" * 220)
```
